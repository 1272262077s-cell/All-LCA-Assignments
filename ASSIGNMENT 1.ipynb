```python
#Creating dictionary
sdictionary = {
    101: "Sakshi", 
    102: "Kunal" ,
    103: "Sunil" ,
    104: "Gauri" 
}
print(sdictionary)
#Add
sdictionary[105] = "Jai"
print("\nAfter adding a student")
print(sdictionary)

#Update
sdictionary[101]= "Sakshi kanjwani"
print ("\nAfter updating the list")
print(sdictionary)

#Delete 
del sdictionary[102]
print ("\nAfter deleting a student")
print (sdictionary)
```
