```python
#reverse the number
num = int(input("Enter a number:"))
reverse = 0
while num> 0:
    digit = num%10
    reverse= reverse*10+digit
    num = num//10
print("Reverse=", reverse)
```

```python
#check whether the number is leap year
year= int(input("Enter a year:"))

if year % 400==0:
    print("The year is a leap year")
elif year % 100==0:
    print("The year is not a leap year")
elif year % 4==0:
    print("The year is a leap year")
else:
    print("The year is not a leap year")
    
    
```
